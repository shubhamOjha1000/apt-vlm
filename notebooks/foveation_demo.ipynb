# Gaze-foveated patching (Steps 1–3) through APT

1. Unit tests (`tests/test_foveation.py`, CPU, no downloads).
2. Demo (`scripts/foveation_demo.py`): config, token-budget table, tokens per image × gaze, overlays.

**Runtime:** any GPU (T4 is fine); CPU also works, just slower.
